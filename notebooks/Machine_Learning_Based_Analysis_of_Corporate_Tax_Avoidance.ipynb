# Machine Learning-Based Analysis of Corporate Tax Avoidance: Evidence from Korean Listed Companies

## Introduction

Corporate tax avoidance has attracted increasing attention from researchers, regulators, and investors because of its direct influence on government tax revenues, corporate transparency, and the fairness of taxation systems. Although tax avoidance generally refers to legal practices used to reduce corporate tax liabilities, aggressive tax planning may create concerns regarding financial reporting quality, corporate governance, and long-term fiscal sustainability. As a result, understanding corporate tax avoidance has become an important topic in accounting, finance, and taxation research.

Recent advances in data analytics and machine learning have provided researchers with new opportunities to analyze corporate tax behaviour using large-scale financial data. Machine learning algorithms can capture nonlinear relationships and interactions among financial variables that may be difficult to identify using traditional statistical approaches. The increasing availability of standardized corporate panel datasets has also made it possible to evaluate different predictive and analytical methods within a consistent research setting.

Despite these developments, several challenges remain. Many previous studies have focused on a single analytical task, such as predicting future tax avoidance or classifying firms into tax avoidance categories. Other studies have evaluated only a limited number of machine learning algorithms or relied on random data splitting methods that may introduce temporal information leakage when panel data are analyzed. In addition, the increasing use of complex machine learning models creates a need for appropriate explainability methods to understand the variables contributing to model predictions.

To address these limitations, this study develops a comprehensive machine learning framework using the Korean Tax Avoidance Panel (KoTaP) dataset as the primary empirical case study. The framework combines descriptive analysis, regression, classification, clustering, hyperparameter optimization, and explainable artificial intelligence within a leakage-free experimental setting. Regression models are used to predict future corporate tax avoidance indicators, while classification models are used to categorize firms according to different levels of tax avoidance risk. Clustering techniques are further applied to identify groups of firms with similar financial characteristics, while SHAP analysis is used to interpret the contribution of important variables to the predictions of the selected machine learning model.

Overall, the study aims to provide a systematic evaluation of different machine learning approaches for corporate tax avoidance analysis using standardized Korean panel data. By applying a consistent leakage-free methodology across the analytical tasks, the study provides a more reliable assessment of predictive performance and offers interpretable insights into the financial factors associated with corporate tax avoidance.

## Research Gap

## Theoretical Framework

Corporate tax avoidance is influenced by a combination of financial, operational, and governance-related characteristics. Therefore, using a single analytical technique may not provide a complete view of corporate tax behaviour. This study adopts a data-driven analytical framework that combines complementary machine learning approaches to examine corporate tax avoidance from different perspectives.

The first perspective is predictive analysis. Regression models are used to estimate future corporate tax avoidance indicators based on information available in the current period. This provides a continuous prediction of future tax avoidance behaviour and allows the study to evaluate how well current firm characteristics explain future outcomes.

The second perspective is risk classification. Classification models are used to assign firms to different tax-risk categories according to their future tax avoidance behaviour. Comparing several classification algorithms provides an empirical basis for identifying the model that offers the most suitable predictive performance for the dataset.

The third perspective is firm heterogeneity. Clustering techniques are used to identify groups of firms with similar financial characteristics without relying on predefined classes. This provides an additional view of corporate tax behaviour and helps examine whether distinct groups of firms can be identified within the Korean corporate environment.

Model optimization and explainability provide two additional dimensions to the framework. Hyperparameter optimization is used to examine whether model performance can be improved through appropriate parameter selection. SHAP-based analysis is then used for the selected machine learning model to examine the contribution of individual financial variables to its predictions.

All predictive analyses are conducted using a leakage-free temporal design. The models are trained using information available before the prediction period and evaluated on later observations. This design is intended to provide a more realistic assessment of the models' ability to predict future corporate tax avoidance.

Overall, the framework combines prediction, classification, firm segmentation, model optimization, and explainability to provide a broader analytical assessment of corporate tax avoidance using the KoTaP dataset.

## Methodology

This study adopts a quantitative research design to investigate corporate tax avoidance using multiple machine learning techniques. The proposed analytical framework integrates descriptive statistics, exploratory data analysis, regression, classification, clustering, hyperparameter optimization, and explainable artificial intelligence.

The research process consists of several sequential stages. First, descriptive and exploratory analyses are conducted to examine the characteristics of the Korean Tax Avoidance Panel (KoTaP) dataset and to explore relationships among financial, governance, and tax-related variables.

Second, data preprocessing and feature engineering procedures are applied to prepare the dataset for analysis. These procedures include handling missing values, transforming variables where necessary, and applying feature scaling to models that require it. To prevent information leakage, preprocessing parameters are estimated using only the relevant training data in predictive experiments.

Third, regression models are developed to predict future corporate tax avoidance indicators using current-period firm information. The regression task follows a one-year-ahead forecasting design, where information available in the current period is used to predict corporate tax avoidance in the subsequent period.

Fourth, classification models are developed to categorize firms into different levels of corporate tax avoidance risk based on their future tax avoidance behaviour. Multiple classification algorithms are evaluated using consistent performance metrics to enable a fair comparison of their predictive capabilities.

Fifth, unsupervised clustering techniques are applied to identify naturally occurring groups of firms based on their financial and tax-related characteristics. Unlike the supervised learning tasks, clustering does not rely on predefined target labels and is used to examine potential heterogeneity among firms.

Hyperparameter optimization is subsequently performed for selected machine learning models to determine whether appropriate parameter configurations can improve predictive performance. The optimized models are then compared with their corresponding baseline models using the same evaluation criteria.

To provide a realistic assessment of predictive performance, the supervised learning experiments follow a leakage-free temporal evaluation protocol. The data are divided chronologically so that earlier observations are used for model development and later observations are reserved for evaluation. In addition, preprocessing and feature transformation procedures are fitted exclusively using the training data before being applied to subsequent periods.

Finally, SHAP (SHapley Additive exPlanations) is employed to interpret the predictions of the selected machine learning model. The analysis identifies the contribution of individual financial variables to model predictions and provides additional insight into the financial characteristics associated with corporate tax avoidance.

Overall, the methodology provides an integrated framework for examining corporate tax avoidance from predictive, classification, segmentation, optimization, and explainability perspectives using the KoTaP dataset.

### Outliers Analysis
Examines the distribution of numerical variables and identifies potential Identifies potential outliers across all numerical variables using the interquartile range (IQR) method.

```python
outlier_summary = []

numeric_columns = df.select_dtypes(include=["int64", "float64"]).columns

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = ((df[col] < lower_bound) | (df[col] > upper_bound)).sum()

    outlier_summary.append([
        col,
        outliers,
        round((outliers / len(df)) * 100, 2)
    ])

outlier_df = pd.DataFrame(
    outlier_summary,
    columns=["Variable", "Outlier_Count", "Outlier_Percentage"]
)

outlier_df.sort_values(
    by="Outlier_Percentage",
    ascending=False
).reset_index(drop=True)
```

**Observation:** The IQR method identifies outliers in several financial variables, particularly sales, assets, liabilities, inventory, and operating cash flow. These observations are expected in corporate financial datasets because firms differ substantially in size and financial performance. Therefore, no observations are removed at this stage to preserve the economic characteristics of the sample.

### Optimal Number of Clusters Selection

Determines the most appropriate number of clusters by evaluating clustering performance across different values of K.

#### Applying the Elbow Method

Evaluates the within-cluster sum of squares (WCSS) across different cluster numbers to identify the point where additional clusters provide diminishing improvements.

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

wcss = []

for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=20
    )

    kmeans.fit(X_clean)
    wcss.append(kmeans.inertia_)

plt.figure(figsize=(8,5))
plt.plot(range(2,11), wcss, marker='o')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('WCSS')
plt.title('Elbow Method (Cleaned Dataset)')
plt.grid(True)
plt.show()
```

**Observation:** The Elbow Method revealed a substantial reduction in WCSS between K = 2 and K = 5, after which the rate of improvement became considerably smaller. This pattern suggests that the optimal number of clusters is likely located within this range.

#### Calculating Silhouette Scores

Measures cluster separation and cohesion to identify the number of clusters that produces the highest clustering quality.

```python
from sklearn.metrics import silhouette_score

silhouette_scores = []

for k in range(2,11):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=20
    )

    labels = kmeans.fit_predict(X_clean)

    score = silhouette_score(
        X_clean,
        labels
    )

    silhouette_scores.append(score)

silhouette_df = pd.DataFrame({
    'K': range(2,11),
    'Silhouette Score': silhouette_scores
})

silhouette_df
```

**Observation:** The highest silhouette score was obtained at K = 2, with clustering quality gradually declining as additional clusters were introduced. Although K = 2 achieved the strongest separation, larger values of K were also considered to explore potential subgroup structures within the dataset.

#### Visualizing Silhouette Scores

Visualizes clustering quality across different values of K to support cluster selection.

```python
plt.figure(figsize=(8,5))

plt.plot(
    range(2,11),
    silhouette_scores,
    marker='o'
)

plt.xlabel('Number of Clusters (K)')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Analysis (Cleaned Dataset)')
plt.grid(True)

plt.show()
```

**Observation:** The silhouette analysis confirms that clustering quality decreases as the number of clusters increases. This result suggests that the dataset exhibits limited natural segmentation and that firms share relatively similar characteristics.

## Gaussian Mixture Model (GMM) Clustering

Applies a probabilistic clustering approach to identify latent firm groups while allowing observations to exhibit overlapping cluster memberships. Unlike K-Means, GMM models clusters as probability distributions, making it more suitable for datasets where cluster boundaries are not clearly separated.

### Gaussian Mixture Model Construction

Builds Gaussian Mixture Models with different numbers of clusters to identify the most appropriate clustering solution.

#### Evaluating Alternative Cluster Solutions

```python
from sklearn.mixture import GaussianMixture
import pandas as pd

bic_scores = []
aic_scores = []

for k in range(2, 11):

    gmm = GaussianMixture(
        n_components=k,
        covariance_type='full',
        random_state=42
    )

    gmm.fit(X_clean)

    bic_scores.append(gmm.bic(X_clean))
    aic_scores.append(gmm.aic(X_clean))

gmm_results = pd.DataFrame({
    'K': range(2,11),
    'BIC': bic_scores,
    'AIC': aic_scores
})

gmm_results
```

## Predictive Modeling Framework

This study adopts a leakage-free one-year-ahead prediction framework to forecast future corporate tax avoidance using firms' financial and governance characteristics. The prediction task is formulated by using explanatory variables observed in year t to predict the corporate tax avoidance measure in year t + 1. This experimental design reflects practical decision-making scenarios in which only currently available information can be used to forecast future tax avoidance behavior while preventing information leakage from future observations.

To provide a comprehensive evaluation, the proposed framework combines traditional machine learning and deep learning techniques. Three regression models (Multiple Linear Regression, LightGBM, and Residual MLP) are employed to predict the future level of corporate tax avoidance, while two classification models (Random Forest and Support Vector Machine) are developed to classify firms according to their tax avoidance risk. All models are evaluated using the same validation strategy and performance metrics to ensure a fair and consistent comparison.

### Constructing the Classification Labels

Constructs dynamic tax-risk classes using year-specific quartiles of the one-year-ahead CETR. Firms in the lowest quartile are labeled as High risk, firms in the highest quartile as Low risk, and the remaining firms as Medium risk. Computing quartiles separately for each fiscal year accounts for annual variation in tax avoidance behavior and avoids arbitrary fixed thresholds.

```python
predict_df['Q1_threshold'] = (
    predict_df.groupby('year')['CETR_t1']
              .transform(lambda x: x.quantile(0.25))
)

predict_df['Q3_threshold'] = (
    predict_df.groupby('year')['CETR_t1']
              .transform(lambda x: x.quantile(0.75))
)

predict_df['TaxRisk'] = 'Medium'

predict_df.loc[
    predict_df['CETR_t1'] <= predict_df['Q1_threshold'],
    'TaxRisk'
] = 'High'

predict_df.loc[
    predict_df['CETR_t1'] >= predict_df['Q3_threshold'],
    'TaxRisk'
] = 'Low'

predict_df.drop(
    columns=['Q1_threshold', 'Q3_threshold'],
    inplace=True
)

predict_df[['year', 'CETR_t1', 'TaxRisk']].head(15)
```

## LightGBM Regression

Develops a LightGBM regression model to predict the one-year-ahead corporate tax avoidance measure (CETR_t1). Hyperparameters are optimized using RandomizedSearchCV with a six-fold expanding window cross-validation strategy. Since LightGBM is a tree-based algorithm, feature scaling is not applied.

### Import Libraries

Imports the required libraries for LightGBM model development, hyperparameter optimization, and regression performance evaluation.

```python
!pip install optuna
import optuna
import lightgbm as lgb
import numpy as np
import pandas as pd
```

### Initialize the Base Model

Initializes the LightGBM regression model using a fixed random seed to ensure reproducibility throughout the optimization process.

```python
base_model = lgb.LGBMRegressor(
    random_state=42,
    verbose=-1
)
```

### Define the Objective Function

Defines the Optuna objective function that evaluates candidate hyperparameter combinations using the six-fold expanding window cross-validation strategy. For each trial, a LightGBM model is trained and validated across all expanding folds, and the average validation RMSE is returned as the optimization objective.

```python
def objective(trial):

    params = {

        'n_estimators': trial.suggest_int('n_estimators', 200, 800),

        'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.10),

        'max_depth': trial.suggest_int('max_depth', 3, 10),

        'num_leaves': trial.suggest_int('num_leaves', 15, 127),

        'min_child_samples': trial.suggest_int('min_child_samples', 10, 50),

        'subsample': trial.suggest_float('subsample', 0.7, 1.0),

        'colsample_bytree': trial.suggest_float('colsample_bytree', 0.7, 1.0),

        'random_state': 42,

        'verbose': -1

    }

    rmse_scores = []

    for train_idx, valid_idx in cv_folds:

        X_train_fold = X_train.loc[train_idx]
        X_valid_fold = X_train.loc[valid_idx]

        y_train_fold = y_train_reg.loc[train_idx]
        y_valid_fold = y_train_reg.loc[valid_idx]

        model = lgb.LGBMRegressor(**params)

        model.fit(
            X_train_fold,
            y_train_fold
        )

        predictions = model.predict(X_valid_fold)

        rmse = np.sqrt(
            mean_squared_error(
                y_valid_fold,
                predictions
            )
        )

        rmse_scores.append(rmse)

    return np.mean(rmse_scores)
```

### Hyperparameter Optimization

Optimizes the LightGBM hyperparameters using Optuna. The optimization process evaluates multiple candidate configurations and selects the parameter set that minimizes the average validation RMSE across the six expanding validation folds.

```python
study = optuna.create_study(
    direction='minimize',
    study_name='LightGBM_Regression'
)

study.optimize(
    objective,
    n_trials=50,
    show_progress_bar=True
)
```

### Display the Best Hyperparameters

Displays the optimal hyperparameter configuration identified by Optuna together with the corresponding minimum validation RMSE.

```python
print("Best RMSE:", study.best_value)

print("\nBest Hyperparameters:")

for key, value in study.best_params.items():
    print(f"{key}: {value}")
```

### Train the Final LightGBM Model

Trains the final LightGBM regression model using the optimal hyperparameters identified by Optuna and the complete training dataset (2011–2022).

```python
final_lgb_model = lgb.LGBMRegressor(
    **study.best_params,
    random_state=42,
    verbose=-1
)

final_lgb_model.fit(
    X_train,
    y_train_reg
)
```

For investors and financial analysts, the classification and regression models provide complementary information. Regression estimates provide a continuous forecast of future CETR, while classification models provide a more interpretable categorization of firms into tax-risk groups. These outputs can potentially be incorporated into broader financial and risk-analysis processes.

The clustering results also provide a useful basis for firm segmentation. Firms can be examined according to their financial profiles rather than relying exclusively on predefined industry or accounting categories. In particular, differences in firm size, foreign ownership, leverage, cash holdings, and asset structure can help analysts understand the heterogeneity of firms within the Korean listed-company market.

Finally, SHAP explainability improves the practical interpretability of the machine learning framework. Instead of relying only on model performance metrics, analysts can examine which financial and tax-related variables contribute most strongly to individual predictions. This can support more transparent analytical interpretation and reduce the gap between predictive modelling and financial reasoning.

##  Results

This section presents the empirical findings obtained from the descriptive analysis, regression models, classification models, clustering techniques, hyperparameter optimization, and SHAP-based explainability analysis. All predictive results are based on the leakage-free temporal framework described in the methodology, where information from the current period is used to predict subsequent corporate tax avoidance behaviour.

### Descriptive and Exploratory Results

The KoTaP dataset contains 12,653 firm-year observations covering 1,754 Korean listed non-financial firms over the period from 2011 to 2024. The dataset includes 65 variables covering financial characteristics, corporate governance, firm structure, and multiple corporate tax avoidance measures.

The initial data quality assessment indicated that the dataset contained no missing values or duplicate observations. Although several financial variables exhibited substantial outliers and skewed distributions, these observations were retained because they represent genuine differences in firm size, financial structure, and operating performance rather than apparent data-entry errors.

The descriptive analysis also revealed substantial variation in corporate tax avoidance indicators across firms and years. The average CETR varied across the study period, reaching relatively high levels during several years and declining in later periods. The yearly analysis showed that CETR and GETR did not follow identical patterns, highlighting the importance of considering both cash-based and accounting-based tax measures when examining corporate tax behaviour.

The exploratory analysis further indicated meaningful relationships between corporate tax avoidance and several firm characteristics. CETR showed a negative correlation with ROA, ROE, and GRW, while positive associations were observed with several historical tax-related variables, firm size, and leverage. These results suggest that profitability, growth, firm characteristics, and historical tax behaviour are relevant dimensions for understanding variation in corporate tax avoidance.

The quartile analysis also showed differences in tax avoidance measures across firm characteristics. In particular, larger firms generally exhibited higher average CETR and GETR values than smaller firms. For example, the average CETR increased from 0.1921 for the smallest size quartile to 0.2755 for the largest quartile. Similarly, firms in higher leverage quartiles generally exhibited higher tax-rate measures than firms in the lowest leverage group.

Overall, the descriptive and exploratory analysis demonstrates substantial heterogeneity in corporate tax behaviour across Korean listed firms and provides support for the subsequent predictive and unsupervised analyses.

### Clustering Results

The clustering analysis was conducted to identify groups of firms with similar financial and governance characteristics. Because the original dataset contained extreme observations that strongly affected distance-based clustering, Isolation Forest was applied only within the clustering workflow. A total of 127 observations were identified as potential outliers, leaving 12,526 observations for clustering.

K-Means analysis indicated that the dataset contains a dominant group of firms with relatively similar characteristics, while a smaller number of observations form more distinct groups. The clustering results suggest that the Korean firms in the sample are not naturally separated into strongly distinct financial segments.

Gaussian Mixture Model analysis provided additional information by allowing overlapping cluster memberships. A four-cluster solution was examined to interpret differences in firm characteristics. The resulting profiles revealed meaningful differences in firm size, foreign ownership, leverage, cash holdings, liquidity, growth, and asset structure.

The largest and most resource-intensive cluster was characterized by high firm size, foreign ownership, leverage, cash holdings, and tangibility. Another cluster contained relatively large firms with strong governance characteristics, while other groups represented smaller firms or firms with higher liquidity and growth characteristics.

Overall, the clustering results indicate that firm size, foreign ownership, cash holdings, and asset structure are important dimensions of firm heterogeneity in the dataset.

### Regression Results

The regression analysis evaluates the ability of machine learning models to predict one-year-ahead CETR using information available in the current period. The predictive dataset contains 10,899 firm-year observations after removing the final observation for each firm, for which a subsequent-year CETR value is unavailable.

Multiple Linear Regression was first used as the statistical baseline. The model achieved an MAE of 0.1753, an RMSE of 0.2320, and an R\textsuperscript{2} of -0.0983. The negative R\textsuperscript{2} indicates that a purely linear specification was not sufficient to explain future CETR variation under the temporal forecasting setting.

LightGBM produced progressively better results as additional explanatory variables were introduced. Using the initial 17-feature configuration, the model achieved an MAE of 0.1724, an RMSE of 0.2229, and an R\textsuperscript{2} of -0.0145. Expanding the feature set to 36 variables improved the results to an MAE of 0.1687, an RMSE of 0.2194, and an R\textsuperscript{2} of 0.0176. The full financial feature configuration further reduced the MAE to 0.1651 and the RMSE to 0.2178 while increasing R\textsuperscript{2} to 0.0315.

The largest improvement was obtained after incorporating tax-history variables that were observable at the prediction period. The optimized LightGBM model achieved a final MAE of 0.1494, an RMSE of 0.2094, and an R\textsuperscript{2} of 0.1051 on the independent test period. These results indicate that historical tax-related information provides important predictive signals for future corporate tax avoidance when combined with nonlinear machine learning methods.

The Residual Multi-Layer Perceptron achieved an MAE of 0.1606, an RMSE of 0.2291, and an R\textsuperscript{2} of -0.0718. Although the neural network improved upon the MLR baseline in terms of MAE, it remained inferior to the optimized LightGBM model.

Overall, the regression experiments demonstrate that feature enrichment and nonlinear modelling substantially improved predictive performance. The optimized LightGBM model achieved the lowest prediction error and the highest R\textsuperscript{2} among the evaluated regression models.

### Classification Results

The classification task categorizes firms into three corporate tax avoidance risk levels: High, Medium, and Low. The classification models were evaluated using Accuracy, Precision, Recall, F1-score, and weighted multiclass ROC-AUC.

Logistic Regression achieved an accuracy of 0.5500, a weighted precision of 0.5300, a weighted recall of 0.5500, a weighted F1-score of 0.4800, and a weighted ROC-AUC of 0.6745. The model performed particularly well in identifying the Medium-risk class, while its ability to identify Low-risk firms was limited.

The baseline Random Forest model improved upon Logistic Regression, achieving an accuracy of 0.5800 and a weighted ROC-AUC of 0.7068. After hyperparameter optimization, the Random Forest model achieved an accuracy of 0.5899, precision of 0.5722, recall of 0.5899, F1-score of 0.5640, and weighted ROC-AUC of 0.7242.

The improvement was consistent across all major evaluation metrics. Accuracy increased by 0.0099, precision by 0.0122, recall by 0.0099, F1-score by 0.0340, and weighted ROC-AUC by 0.0174.

The baseline SVM achieved an accuracy of 0.5800, precision of 0.6600, recall of 0.5800, F1-score of 0.5000, and weighted ROC-AUC of 0.7203. However, hyperparameter optimization reduced most of its evaluation metrics. The optimized SVM achieved an accuracy of 0.5608, precision of 0.5276, recall of 0.5608, F1-score of 0.5092, and weighted ROC-AUC of 0.6666.

Therefore, the optimized Random Forest provided the strongest overall balance among the evaluated classification models. The results also demonstrate that hyperparameter optimization does not necessarily improve every algorithm, as the effect depends on the model and the characteristics of the dataset.

### SHAP Explainability Results

SHAP analysis was applied to the final optimized LightGBM regression model to interpret the contribution of individual variables to one-year-ahead CETR predictions.

The global SHAP analysis identified A\_GETR as the most influential explanatory variable, with a mean absolute SHAP value of 0.0311. It was followed by GETR5 with a mean SHAP value of 0.0242, PTI with 0.0103, CETR5 with 0.0098, and TSDA with 0.0082. Other important variables included ROE, GRW, lagged net income, A\_CETR, and PPE.

The dominance of historical tax-related variables indicates that previous tax behaviour contains substantial information about future corporate tax avoidance. The SHAP dependence analysis further shows a nonlinear relationship between A\_GETR and the model output, supporting the earlier finding that nonlinear machine learning models are able to capture relationships that are difficult to represent through a simple linear specification.

The SHAP results therefore complement the predictive evaluation by identifying the financial and tax-related variables that contribute most strongly to the model's one-year-ahead predictions.

## Discussion

The findings provide several important insights into the use of machine learning for corporate tax avoidance analysis. The results indicate that corporate tax avoidance is not easily explained through a simple linear relationship with firm characteristics. The negative R\textsuperscript{2} obtained from the Multiple Linear Regression baseline suggests that linear modelling alone does not adequately capture the complexity of future CETR behaviour.

The stronger performance of LightGBM demonstrates the value of nonlinear modelling for this problem. LightGBM consistently improved as the feature space was expanded, and the largest improvement occurred after the inclusion of historical tax-related variables. This result suggests that corporate tax behaviour contains an important temporal component, where previous tax-related outcomes provide information about subsequent corporate tax behaviour.

An important finding is that increasing model complexity alone did not guarantee better performance. The Residual MLP performed worse than the optimized LightGBM despite being capable of modelling nonlinear relationships. This suggests that the quality and relevance of the input variables may be more important than simply using a more complex model. In this study, systematic feature engineering and the inclusion of informative tax-history variables contributed more to predictive performance than the use of a neural-network architecture.

The classification results provide a similar perspective. Nonlinear ensemble learning improved upon the Logistic Regression baseline, with the optimized Random Forest achieving the strongest overall classification performance. However, the improvement in accuracy was relatively modest. This indicates that future corporate tax-risk categories cannot be perfectly determined from the available financial characteristics. The moderate predictive performance also suggests that corporate tax behaviour may depend on additional factors that are not fully represented in the available variables.

The classification results further show that the models had greater difficulty identifying the Low-risk category, while the Medium-risk category was identified more successfully. This pattern is partly related to the distribution of the target categories, where the Medium category contains approximately half of the observations. Therefore, overall accuracy should not be interpreted independently from class-specific metrics.

The contrasting effects of hyperparameter optimization are also noteworthy. Optimization improved the Random Forest model across all major metrics but reduced the performance of the SVM model. This finding demonstrates that hyperparameter optimization should be treated as an empirical procedure rather than an assumption that optimization will automatically improve every model. Model-specific characteristics and the structure of the data influence whether a more complex parameter configuration is beneficial.

The clustering analysis provides additional evidence of firm heterogeneity but also indicates that the sample does not exhibit extremely strong natural segmentation. The GMM profiles show meaningful differences among groups, particularly in terms of firm size, foreign ownership, leverage, cash holdings, and asset structure. However, the existence of a dominant group and smaller clusters suggests that most firms share a relatively similar overall financial structure, while only a smaller proportion display more distinctive characteristics.

The SHAP analysis provides an additional interpretation of the regression results. The high importance of A\_GETR and GETR5 indicates that historical tax measures are among the strongest predictors of future CETR. This finding is consistent with the substantial improvement observed when tax-history variables were incorporated into the LightGBM model. The SHAP results therefore strengthen the argument that corporate tax avoidance exhibits persistence over time.

From a methodological perspective, the study also demonstrates the importance of temporal validation when analysing corporate panel data. By using chronological training and testing periods and restricting preprocessing to information available during the training period, the study provides a more realistic evaluation of future predictive performance than a conventional random split would provide.

Nevertheless, the results should be interpreted with caution. The final R\textsuperscript{2} of approximately 0.105 indicates that the optimized LightGBM model explains only a limited proportion of the variation in future CETR. Therefore, the model should not be interpreted as a complete explanation of corporate tax avoidance. Instead, the results indicate that machine learning can identify useful predictive patterns while substantial unexplained variation remains.

The findings also highlight the importance of distinguishing prediction from causal interpretation. SHAP identifies the contribution of variables to model predictions, but these contributions should not be interpreted as evidence that the corresponding variables causally determine corporate tax avoidance. Further econometric and institutional analysis would be required to establish causal relationships.

## Conclusion

This study examined corporate tax avoidance among Korean listed companies using a comprehensive machine learning framework based on the KoTaP dataset. The results show that nonlinear machine learning models, particularly the optimized LightGBM model, can provide better one-year-ahead CETR predictions than the linear baseline. The optimized Random Forest also achieved the strongest overall classification performance, while clustering revealed meaningful differences in firms' financial characteristics.

SHAP analysis further showed that historical tax-related variables, particularly A\_GETR and GETR5, were among the most important contributors to future CETR predictions. Overall, the findings demonstrate that combining prediction, classification, clustering, and explainability can provide a broader understanding of corporate tax avoidance.

However, the moderate predictive performance indicates that corporate tax behaviour is influenced by factors beyond the available financial variables. Future research could therefore incorporate additional governance, institutional, industry, and macroeconomic factors and evaluate the framework using other datasets and forecasting settings.
